# 02 - Exploratory Data Analysis

Questions, summaries, and reusable Step 02 figures. All analysis reads processed Step 01 files; EDA fields are temporary.

## 1. Imports
Question: can this EDA be reproduced from the project root?

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display
from src.data.eda import add_eda_features, transportation_rate
ROOT = Path.cwd().resolve()
if not (ROOT / 'data').exists(): ROOT = ROOT.parent
ASSETS = ROOT / 'web/assets/generated'
```

## 2. Load processed data
Question: what data are available for target-aware EDA?

```python
train = pd.read_csv(ROOT / 'data/processed/train_clean.csv')
test = pd.read_csv(ROOT / 'data/processed/test_clean.csv')
eda = add_eda_features(train)
{'train_shape': train.shape, 'test_shape': test.shape, 'train_missing': int(train.isna().sum().sum())}
```

## 3. Dataset overview and target analysis
Question: is the target sufficiently balanced for descriptive comparisons?

```python
train.Transported.value_counts().rename_axis('Transported').to_frame('passengers').assign(percent=lambda x: x.passengers / len(train) * 100)
display(Image(filename=str(ASSETS / 'target_distribution.png')))
display(Image(filename=str(ASSETS / 'missing_values_overview.png')))
```

## 4. Age analysis
Question: does the observed target rate differ across ages?

```python
train.Age.describe()
display(Image(filename=str(ASSETS / 'age_distribution.png')))
display(Image(filename=str(ASSETS / 'age_by_transportation.png')))
```

## 5. Home planet and destination
Question: how do passenger origin and destination compare by observed outcome?

```python
transportation_rate(eda, 'HomePlanet'), transportation_rate(eda, 'Destination')
display(Image(filename=str(ASSETS / 'homeplanet_distribution.png')))
display(Image(filename=str(ASSETS / 'homeplanet_vs_transportation.png')))
display(Image(filename=str(ASSETS / 'destination_distribution.png')))
display(Image(filename=str(ASSETS / 'destination_vs_transportation.png')))
```

## 6. CryoSleep and VIP
Question: what observed target contrasts appear in these passenger-status fields?

```python
transportation_rate(eda, 'CryoSleep'), transportation_rate(eda, 'VIP')
display(Image(filename=str(ASSETS / 'cryosleep_distribution.png')))
display(Image(filename=str(ASSETS / 'cryosleep_vs_transportation.png')))
display(Image(filename=str(ASSETS / 'vip_distribution.png')))
display(Image(filename=str(ASSETS / 'vip_vs_transportation.png')))
```

## 7. Spending analysis
Question: are the spending variables skewed, zero-inflated, or outcome-associated?

```python
eda[['RoomService','FoodCourt','ShoppingMall','Spa','VRDeck','TotalSpending']].describe().T
display(Image(filename=str(ASSETS / 'spending_distributions.png')))
display(Image(filename=str(ASSETS / 'total_spending_vs_transportation.png')))
display(Image(filename=str(ASSETS / 'spending_by_homeplanet.png')))
```

## 8. Cabin analysis
Question: do temporary cabin deck and side fields show descriptive differences?

```python
transportation_rate(eda, 'CabinDeck'), transportation_rate(eda, 'CabinSide')
display(Image(filename=str(ASSETS / 'cabin_deck_distribution.png')))
display(Image(filename=str(ASSETS / 'cabin_deck_vs_transportation.png')))
display(Image(filename=str(ASSETS / 'cabin_side_vs_transportation.png')))
```

## 9. Passenger group analysis
Question: does verified PassengerId grouping reveal common group sizes?

```python
eda.drop_duplicates('GroupId').GroupSize.value_counts().sort_index()
display(Image(filename=str(ASSETS / 'group_size_distribution.png')))
display(Image(filename=str(ASSETS / 'group_size_vs_transportation.png')))
```

## 10. Correlation and train/test checks
Question: are numeric relationships and broad train/test distributions suitable for later preprocessing?

```python
pd.DataFrame({'train_missing_pct': train.isna().mean()*100, 'test_missing_pct': test.isna().mean()*100}).round(2)
{'train_age_mean': train.Age.mean(), 'test_age_mean': test.Age.mean()}
display(Image(filename=str(ASSETS / 'correlation_heatmap.png')))
```
